Mapping Alzheimer's: Linking ABeta42 and pTAU Protein Levels to Cognitive Decline in the Aging Brain

## Project Goal:
This project is all about figuring out how specific physical changes in the brain connect to real-world memory loss. Specifically, I want to see if the levels of two major brain proteins (ABeta42 and pTAU) can actually predict how well a patient performs on cognitive tests while they are alive. The two main questions we seek to answer are:

1. Do ABeta42 protein levels differ between donors who had dementia and those who didn't? Furthermore, is there a direct link between a person's ABeta42 levels and their actual cognitive performance on their last MMSE exam?

2. Do pTAU protein levels differ based on a patient's dementia status? Also, can higher levels of pTAU be tied directly to lower MMSE scores?

* Risk factors (genetic, lifestyle) 
Genetics can play a big role in a person's risk of developing Alzheimer's, especially when looking at the APOE gene, which is one of the main genetic risk factors for the late-onset Alzheimer's. Everyone has two copies of the APOE gene, one from each biological parent, and the three common variants are ε2, ε3, and ε4. The ε2 variant is generally considered protective, ε3 is the most common and is pretty neutral, and ε4 is associated with a higher risk of developing Alzheimer’s. Having two ε4 alleles can increase that risk even more, but it still does not mean that a person will definitely develop the disease. This is because genetics are not the only part that can contribute to Alzheimer's; lifestyle and environmental factors can also affect a person's risk and overall brain health. Things like not being physically active, smoking, poor cardiovascular health, diet, and social isolation can all have an impact over time. While a person cannot change their genetics, they can change their lifestyle factors, which is why staying physically active, taking care of cardiovascular health, and staying socially connected can be important for maintaining brain health.

* Diagnosis
Diagnosing Alzheimer's disease usually involves multiple tests rather than one single test. Doctors will first look at a patient's medical history and ask the patient or their family about changes in memory, behavior, personality, and their ability to complete everyday activities. They may also use cognitive tests that evaluate things like memory, problem-solving, attention, and language. Blood and urine tests can help rule out other conditions that could be causing similar symptoms, while brain imaging such as MRI, CT, or PET scans can be used to look for changes in the brain or eliminate other possible causes. Biomarker testing can provide additional evidence of Alzheimer's-related changes, including tests of cerebrospinal fluid (CSF) for proteins associated with the disease and certain PET scans that detect abnormal proteins in the brain. However, biomarkers are generally considered alongside the rest of the clinical evaluation rather than being imterpreted completely on their own.

* Disease progression & prognosis
Alzheimer's disease is progressive, meaning that symptoms usually start slowly and get worse over time. In the earlier stages, a person may experience mild memory loss or difficulty with things like problem-solving and completing familiar tasks, but many people are still able to stay fairly independent. As the disease progresses into the moderate stages, memory and thinking problems become more noticable, and a person may need more help with everyday activities. In the later stages, Alzheimer's can affect communication, movement, and the ability to care for oneself, making around-the-clock assistance necessary. The rate of progression is different for everyone, and the disease can develop over many years, so if can be difficult to predict exactly how quickly someone;s symptoms will change.

* Continuum of care providers
The continuum of care for someone with Alzheimer's disease usually starts with a primary care provider, who may notice memory or cognitive changes, perform initial assessments, and refer the patient to specialists. A neurologist or geriatrician may then perform a more detailed evaluation and help diagnose and manage the disease, while a radiologist may be involved in interpreting brain imaging used during diagnosis and treatment monitoring. * Biological mechanisms (anatomy, organ physiology, cell & molecular physiology)
Alzheimer's disease affects several areas of the brain that are important for memory, learning, and behavior. The entorhinal cortex and hippocampus are especially important because they are involved in processing and forming memories, and these regions are affected early in Alzheimer's disease. The amygdala is involved in processing emotions, including fear, and changes in connected brain regions can contribute to emotional and behavioral changes as the disease progresses. At the cellular level, Alzheimer's is associated with the buildup of amyloid-beta plaques outside neurons and abnormal tau protein tangles inside neurons. Normally, tau helps stabilize microtubules, which act like an internal transport system that moves nutrients and other matierals through neurons. In Alzheimer's tau becomes abnormally modified and separates from the microtubules, eventually forming neurofibrillary tangles. Along with amyloid buildup, these changes interfere with communication between neurons and contribute to neurons becoming dysfunctional and eventually dying.
(https://pmc.ncbi.nlm.nih.gov/articles/PMC8807768/)

## Data-Set: 
For this analysis, we are using a subset of the SEA-AD Cohort Metadata Dataset, which comes from a partnership between the Seattle Alzheimer's Disease Brain Cell Atlas and the Allen Insitute for Brain Science. The data brings together medical histories from donors who took part in major aging research programs, including the Adult Changes in Thought (ACT) study and the University of Washington Alzheimer's Disease Research Center (UW ADRC). While the donors were alive, doctors tracked their memory and cognitive skills using the Mini-Mental State Examination (MMSE). After they passed away, pathologists collected brain tissue samples and ran specialized laboratory tests to count the exact protein concentrations left behind. Our analysis focuses on:
    - Cognitive status: the primary diagnostic category grouping donors into either Dementia or No dementia.
    - Last MMSE Score: The donor's final score on the Mini-Mental State Examination, which is a standard cognitive test scored on a continuous scale from 0 to 30 (where lower scores show a more severe cognitive impairment)
    - ABeta42 Concentration: The amount of anyloid-beta 42 protein found in the tissue, measured in picograms per microgram (pg/ug) of total protein.
    - pTAU Concentration: The amount of phosphorylated tau protein found in the tissue, also measured in picograms per microgram (pg/ug) of total protein.

Seattle Alzheimer’s Disease Brain Cell Atlas (SEA-AD). (2024). SEA-AD Cohort Donor Metadata [Data set]. Allen Institute for Brain Science.

## Data Analysis: 
Donor records were loaded into Patient objects, storing cognitive status, ABeta42 level, pTAU level, and last MMSE score. Missing values of donors were excluded from specific tests, this caused sample sizes to slightly vary. Two analysis were run for each protein, one analysis was a group comparison which included the mean protein level in donors with dementia versus no dementia, Welch's t-test, which was chosen due to it not assuming equal variances between groups and a Mann-Whitney U test was used to confirm. The second analysis was a regression which used linear regression to estimate the relationship and significance of protein level versus the patients last MMSE score.

Module_1_patient.py:

```python
import csv #import csv module to read CSV files
class Patient: #class to represent a patient with cognitive status, protein levels, and MMSE score
    all_patients = [] 

    def __init__(self, cognitive_status: str, abeta42: float, ptau: float, mmse: float): #defines the constructor for the Patient class, which takes in cognitive status, abeta42 level, ptau level, and MMSE score as arguments
        self.cognitive_status = cognitive_status
        self.abeta42 = abeta42
        self.ptau = ptau
        self.mmse = mmse
        Patient.all_patients.append(self)
    def __repr__(self): #defines the string representation of the Patient object, which returns a string with the cognitive status, abeta42 level, ptau level, and MMSE score of the patient
        return f"({self.cognitive_status} | ABeta42 = {self.abeta42} | pTAU = {self.ptau} | MMSE = {self.mmse})"
        
    @classmethod
    def instantiate_from_csv(cls, filename: str): #class method that reads patient data from a CSV file and creates Patient objects for each row in the file
        
        with open(filename, encoding="utf8") as f: #opens the CSV file with UTF-8 encoding
            reader = csv.DictReader(f)
            rows_of_patients = list(reader)

            for row in rows_of_patients: #iterates through each row in the CSV file and extracts the cognitive status, abeta42 level, ptau level, and MMSE score for each patient. If any of these values are missing, they are set to None. A new Patient object is then created for each row using the extracted values.
                if row['ABeta42 pg/ug'] != '':
                    abeta42 = float(row['ABeta42 pg/ug'])
                else:
                    abeta42 = None
                if row['pTAU pg/ug'] != '':
                    ptau = float(row['pTAU pg/ug'])
                else:
                    ptau = None
                if row['Last MMSE Score'] != '':
                    mmse = float(row['Last MMSE Score'])
                else:
                    mmse = None

                cls(
                    cognitive_status=row['Cognitive Status'],
                    abeta42=abeta42,
                    ptau=ptau,
                    mmse=mmse
                )
    @classmethod
    def filter(cls, patient_list, cognitive_status: str = "any"): #class method that filters a list of Patient objects based on their cognitive status. If the cognitive status is set to "any", all patients are returned. Otherwise, only patients with the specified cognitive status are returned.
        remaining_patients = patient_list
        if cognitive_status != "any":
            remaining_patients = [patient for patient in remaining_patients
                                  if patient.cognitive_status == cognitive_status]
        return remaining_patients
```

Module_1_main.py:

```python
#import necessary libraries
import matplotlib
matplotlib.use('TkAgg')
from module_1_patient import *
import matplotlib.pyplot as plt
from scipy import stats
import numpy as np
import statistics
from sklearn.linear_model import LinearRegression

Patient.instantiate_from_csv(r"C:\Users\ikedo\OneDrive\Desktop\Computational BME\Module 1\BME2315_Module1\Metadata and Protein Data for Module 1 (1).csv") #grabs the CSV file and creates Patient objects for each row in the file

def bar_by_status(attr, label): #defines a function that creates a bar graph comparing the mean protein levels (mean ± SD) for patients with "No dementia" vs "Dementia", with individual donors overlaid. The function also performs Welch's t-test and Mann-Whitney U test to analyze the differences between the two groups.
    """Bar graph of mean protein level (mean +/- SD) for No dementia vs Dementia,
    with individual donors overlaid, analyzed with Welch's t-test and Mann-Whitney U."""
    groups = ["No dementia", "Dementia"]

    values = [] 
    for status in groups: 
        values.append([getattr(p, attr) for p in Patient.filter(Patient.all_patients, cognitive_status=status) #creates a list of protein levels for patients with the specified cognitive status, filtering out any patients with missing values for the specified attribute
                       if getattr(p, attr) is not None])

    means = [statistics.mean(v) for v in values] #calculates the mean protein level for each group
    stdevs = [statistics.stdev(v) for v in values] #calculates the standard deviation of protein levels for each group

    t_stat, p_val = stats.ttest_ind(values[0], values[1], equal_var=False) #Welch's t-test
    u_stat, u_p = stats.mannwhitneyu(values[0], values[1]) #Mann-Whitney U test
    print(f'{label}: n = {len(values[0])} (No dementia), {len(values[1])} (Dementia)') #prints the number of patients in each group
    print(f'{label}: means = {means[0]:.3f}, {means[1]:.3f} | Welch t = {t_stat:.3f}, p = {p_val:.3e} | Mann-Whitney p = {u_p:.3e}') #prints the means of each group, Welch's t-test statistic and p-value, and Mann-Whitney U test p-value

    x_labels = [f"{groups[i]}\n(n={len(values[i])})" for i in range(2)]
    plt.bar(x_labels, means, yerr=stdevs, capsize=10, color=["blue", "orange"], alpha=0.6)

    for i in range(2): #iterates through each group and adds individual data points to the bar graph with a small random jitter to avoid overlap
        jitter = np.random.uniform(-0.15, 0.15, len(values[i]))
        plt.scatter(i + jitter, values[i], color="black", s=10, zorder=3)
    
    top = max(max(v) for v in values)
    plt.ylim(0, top * 1.25) #sets the y-axis limit to 1.25 times the maximum protein level across both groups to ensure that all data points and error bars are visible
    plt.text(0.5, top * 1.05, f"Welch t = {t_stat:.2f}\np = {p_val:.3e}", ha='center', va='bottom') #adds a text box to the bar graph displaying the Welch's t-test statistic and p-value
    plt.title(f"Average {label} by Cognitive Status (mean ± SD)") #adds a title to the bar graph indicating that it shows the average protein levels by cognitive status, along with the mean and standard deviation
    plt.xlabel("Cognitive Status") #adds a label to the x-axis indicating that it represents cognitive status
    plt.ylabel(f"{label} (pg/ug)") #adds a label to the y-axis indicating that it represents protein levels in pg/ug
    plt.show() #displays the bar graph

def scatter_vs_mmse(attr, label): #defines a function that creates a scatter plot of protein level vs MMSE score with a linear regression line, equation, R-squared and p-value. The function also prints the slope, R-squared and p-value of the linear regression.
    """Scatter plot of protein level vs MMSE score with a linear regression line,
    equation, R-squared and p-value."""
    mmse_list = []
    level_list = []
    for patient in Patient.all_patients:
        if patient.mmse is not None and getattr(patient, attr) is not None:
            mmse_list.append(patient.mmse)
            level_list.append(getattr(patient, attr))

    X = np.array(mmse_list).reshape(-1, 1)  # Reshape for sklearn
    y = np.array(level_list)
    model = LinearRegression()
    model.fit(X, y)

    slope = model.coef_[0]
    intercept = model.intercept_
    r2 = model.score(X, y)

    p_value = stats.linregress(mmse_list, level_list).pvalue
    print(f'{label} vs MMSE: n = {len(mmse_list)}, slope = {slope:.3f}, R² = {r2:.3f}, p = {p_value:.3e}')

    equation = f"y = {slope:.2f}x + {intercept:.2f}\nR² = {r2:.2f}, p = {p_value:.3f}\nn = {len(mmse_list)}"
    plt.text(0.05, 0.95, equation, color='red', fontsize=12, va='top', transform=plt.gca().transAxes)
    # Create scatter plot with regression line
    plt.scatter(X, y, color='blue')
    plt.plot(X, model.predict(X), color='red')
    plt.xlabel('MMSE Score')
    plt.ylabel(f'{label} (pg/ug)')
    plt.title(f'Scatter Plot of {label} vs MMSE Score')
    plt.show()

bar_by_status("abeta42", "ABeta42") #creates a bar graph comparing the mean ABeta42 protein levels (mean ± SD) for patients with "No dementia" vs "Dementia", with individual donors overlaid. The function also performs Welch's t-test and Mann-Whitney U test to analyze the differences between the two groups.
scatter_vs_mmse("abeta42", "ABeta42") #creates a scatter plot of ABeta42 protein level vs MMSE score with a linear regression line, equation, R-squared and p-value. The function also prints the slope, R-squared and p-value of the linear regression.
bar_by_status("ptau", "pTAU") #creates a bar graph comparing the mean pTAU protein levels (mean ± SD) for patients with "No dementia" vs "Dementia", with individual donors overlaid. The function also performs Welch's t-test and Mann-Whitney U test to analyze the differences between the two groups.
scatter_vs_mmse("ptau", "pTAU") #creates a scatter plot of pTAU protein level vs MMSE score with a linear regression line, equation, R-squared and p-value. The function also prints the slope, R-squared and p-value of the linear regression.
```

## Limitations and Future Work: 
A major limitation in our analysis comes down to the uncertainty around our predictions and the stability of our data. Even though our negative correlation between pTAU and MMSE score shows up as statistically significant, the model has a very low R^2 value of just 0.09, meaning there is a massive 91% of the variance in pTAU levels. This leaves us with a lot of predictive uncertainty. To make our findings much more precise next time, we need to focus heavily on handling extreme outliers and expanding our sample size. Right now, our models are highly sensitive to high-leverage outliers sitting way outside the normal data. Because standard linear regression squares its residuals, these extreme data points exert an abnormal mathematical pull on our trendline. If we eliminate these outliers, our data becomes much more precise, and our resulting p-value shifts drastically because the trendline is no longer being altered by a few anomalous patients. Additionally, while our current sample size lets us find a significant trend, getting a much larger sample size next time will give us a more balanced, more reliable dataset that is more stable against individual variations. 
Moving forward, we can directly address these limitations by changing our data and modeling approach. We can update our instantiate_from_csv workflow to remove extreme biological anomolies using Z-scores, which would clean up the extreme data points without us having to remove them manually. Also, since looking at pTAU alone leaves 91% of the data unexplained, we can change our regression function to include a couple of other columns already in our file, like Age at Death or Years of Education, to see if a multi-variable model gives us a more accurate graph for predicting cognitive decline.
