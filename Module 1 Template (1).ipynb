## Project Goal:
This project is all about figuring out how specific physical changes in the brain connect to real-world memory loss. ## Disease Background: 

* Prevalence & incidence
About 1 in 9 people aged 65 and older (around 11%) has Alzheimer's disease. That means roughly 7.4 million Americans in that age group are living with it right now, and 74% of them are 75 or older. The chances of getting Alzheimer's shoot up as people age, going from 5% for people aged 65-74, to 13% for ages 75-84, and 33% for anyone 85 and older. There are also noticible differences across demographic groups. About 2/3 of Americans living with Alzheimer's are women, while older Black Americans are about twice as likely and older Hispanic Americans are about 1.5 times as likely to have Alzheimer's compared with older White Americans. Although Alzheimer's primarily affects older adults, younger-onset Alzheimer's and other dementias can also affect people under the age of 65. The rate of new cases also increases with age, showing how strongly aging is associated with the development of Alzheimer's symptoms. Globally, tens of millions of people are living with dementia, with millions of new cases occurring each year and Alzheimer's disease accounting for the majority of dementia cases. 
(https://www.alz.org/alzheimers-dementia/facts-figures)

* Economic burden
Alzheimer's disease creates a huge financial burden, not only for the person with the disease but also for their family, caregivers, and the healthcare system. The total economic burden of Alzheimer's and related dementias in the United States is projected to be around $818 billion annually, including healthcare costs and the value of unpaid caregiving. Direct healthcare spending along accounts for around $230 billion each year, with programs like Medicare and Medicaid covering a large portion of those costs. There can also be a loss of income and productivity when people with Alzheimer's are no longer able to work or when family members have to decrease their work hours or leave their jobs to become caregivers. As the disease progresses and a person needs more supervision, home health services, or long-term care, the financial burden increases even more. Globally, the cost of dementia has also been projected to reach around $2 trillion, showing that Alzheimer's is not only a major health issue but also a major economic one. 
(https://www.alz.org/getmedia/bbdf2c43-90c7-4f81-8d50-b49567553592/sdoh-dementia-risk-compiled-report.pdf)(https://www.google.com/url?q=https://academic.oup.com/psychsocgerontology/article/79/12/gbae173/7816889&sa=D&source=docs&ust=1790567378292935&usg=AOvVaw33Y6VzOyJke2sajVd4Ffzx)

* Risk factors (genetic, lifestyle) 
Genetics can play a big role in a person's risk of developing Alzheimer's, especially when looking at the APOE gene, which is one of the main genetic risk factors for the late-onset Alzheimer's. Everyone has two copies of the APOE gene, one from each biological parent, and the three common variants are ε2, ε3, and ε4. The ε2 variant is generally considered protective, ε3 is the most common and is pretty neutral, and ε4 is associated with a higher risk of developing Alzheimer’s. Having two ε4 alleles can increase that risk even more, but it still does not mean that a person will definitely develop the disease. This is because genetics are not the only part that can contribute to Alzheimer's; lifestyle and environmental factors can also affect a person's risk and overall brain health. Things like not being physically active, smoking, poor cardiovascular health, diet, and social isolation can all have an impact over time. While a person cannot change their genetics, they can change their lifestyle factors, which is why staying physically active, taking care of cardiovascular health, and staying socially connected can be important for maintaining brain health. 
(https://www.nia.nih.gov/10-years-alzheimers-disease-and-related-dementias-research/advanced-understanding-dementia-risk) (https://jamanetwork.com/journals/jamaneurology/fullarticle/2811630)

* Societal determinants
- Eductaion: lower levels of education have been associated with poorer brain health and a greater risk of cognitive decline. One possible explanation is cognitive reserve, which is the brain's ability to compensate for age- or disease-related changes. Greater cognitive reserve may allow for someone to tolerate more brain changes before noticeable cognitive symptoms appear. 
- Socioeconomic status/income: lower socioeconomic status can affect access to healthcare and other resources that support brain health and has been associated with a higher risk of dementia. 
    - Healthcare access: having consistent access to quality healthcare can allow for earlier diagnosis, better management of chronic conditions, care planning, and continuted monitoring of Alzheimer's disease. On the other hand, limited healthcare access can delay diagnosis and access to treatment and support.
- Social life: social isolation and limited social connection can negatively affect brain health. Maintaining social connections and engagement is considered an important part of supporting cognitive health as people age.
- Environment: a person's physical environment can also influence brain health. Unsafe or unhealthy housing and neighborhoods with limited opportunities for exercise, transportation, services, and community interaction can create barriers to behaviors that support healthy aging. Communities with things like sidewalks, bike lanes, and accessible services can provide more opportunities for physical activity and social connection.

* Diagnosis
Diagnosing Alzheimer's disease usually involves multiple tests rather than one single test. Doctors will first look at a patient's medical history and ask the patient or their family about changes in memory, behavior, personality, and their ability to complete everyday activities. They may also use cognitive tests that evaluate things like memory, problem-solving, attention, and language. Blood and urine tests can help rule out other conditions that could be causing similar symptoms, while brain imaging such as MRI, CT, or PET scans can be used to look for changes in the brain or eliminate other possible causes. Biomarker testing can provide additional evidence of Alzheimer's-related changes, including tests of cerebrospinal fluid (CSF) for proteins associated with the disease and certain PET scans that detect abnormal proteins in the brain. However, biomarkers are generally considered alongside the rest of the clinical evaluation rather than being imterpreted completely on their own.
(https://www.nia.nih.gov/health/alzheimers-symptoms-and-diagnosis/how-alzheimers-disease-diagnosed)
(https://www.nia.nih.gov/health/alzheimers-symptoms-and-diagnosis/how-biomarkers-help-diagnose-dementia)

* Standard of care treatments (& reimbursement) 
Alzheimer's treatment usually includes a combination of medications and supportive care to help manage symptoms and maintain a person's quality of life. Medications like donepezil, rivastigmine, galantamine, and memantine can help with things like memory, thinking, and being able to complete everyday tasks. There are also newer treatments like lecanemab and donanemab that can be used for some people in the early stages of Alzheimer's to help slow the progress of the disease. Other parts of treatment can include staying physically and socially active, keeping a consistent routine, making the home safer and easier to navigate, and providing support for caregivers. Overall, the goal of treatment is to help people stay as independent and comfortable as possible while managing the symptoms of the disease.
Medicare and Medicaid can help cover many of the treatments used for Alzheimer's disease, although the amount of coverage depends on the treatment and the patient's insurance plan. Medicaid programs provide prescription drug coverage, but specific coverage and reimbursement policies can vary by state. Medicare also covers certain newer Alzheimer's treatments, including anti-amyloid medications, for eligible patients with mild cognitive impairment or mild Alzheimer's dementia when specific coverage requirements are met. For example, Medicare coverage requires the patient to meet the clinical criteria and the treating provider to participate in an approved registry that collects information about treatment outcomes. For lecanemab specifically, patients with Original Medicare generally pay the standard 20% coinsurance after meeting their Part B deductible, although supplemental insurance or Medicare Advantage can change the amount they pay out of pocket. Overall, Medicare and Medicaid can make Alzheimer's treatments more accessible, but coverage and out-of-pocket costs depend on the the patient's specific medication, how eligible they are, and their insurance plan.

* Disease progression & prognosis
Alzheimer's disease is progressive, meaning that symptoms usually start slowly and get worse over time. In the earlier stages, a person may experience mild memory loss or difficulty with things like problem-solving and completing familiar tasks, but many people are still able to stay fairly independent. As the disease progresses into the moderate stages, memory and thinking problems become more noticable, and a person may need more help with everyday activities. In the later stages, Alzheimer's can affect communication, movement, and the ability to care for oneself, making around-the-clock assistance necessary. The rate of progression is different for everyone, and the disease can develop over many years, so if can be difficult to predict exactly how quickly someone;s symptoms will change.
(https://www.mayoclinic.org/diseases-conditions/alzheimers-disease/in-depth/alzheimers-stages/art-20048448)

* Continuum of care providers
The continuum of care for someone with Alzheimer's disease usually starts with a primary care provider, who may notice memory or cognitive changes, perform initial assessments, and refer the patient to specialists. A neurologist or geriatrician may then perform a more detailed evaluation and help diagnose and manage the disease, while a radiologist may be involved in interpreting brain imaging used during diagnosis and treatment monitoring. As Alzheimer's progresses, nurses, occupational/physical therapists, social workers, and other caregivers may become more involved to help with medications, daily activities, mobility, safety, and overall quality of life. (https://www.alzheimers.gov/professionals/health-care-providers)

* Biological mechanisms (anatomy, organ physiology, cell & molecular physiology)
Alzheimer's disease affects several areas of the brain that are important for memory, learning, and behavior. The entorhinal cortex and hippocampus are especially important because they are involved in processing and forming memories, and these regions are affected early in Alzheimer's disease. The amygdala is involved in processing emotions, including fear, and changes in connected brain regions can contribute to emotional and behavioral changes as the disease progresses. At the cellular level, Alzheimer's is associated with the buildup of amyloid-beta plaques outside neurons and abnormal tau protein tangles inside neurons. Normally, tau helps stabilize microtubules, which act like an internal transport system that moves nutrients and other matierals through neurons. In Alzheimer's tau becomes abnormally modified and separates from the microtubules, eventually forming neurofibrillary tangles. Along with amyloid buildup, these changes interfere with communication between neurons and contribute to neurons becoming dysfunctional and eventually dying.
(https://pmc.ncbi.nlm.nih.gov/articles/PMC8807768/)

* Clinical Trials/next-gen therapies
There are currently many clinical trials looking for new ways to treat Alzheimer's disease and slow its progression. In 2026, the Alzheimer's drug-development pipeline included 158 different therapies across 192 clinical trials, showing that researchers are exploring many approaches beyond current available medications. While some newer treatments continue to target amyloid-beta, there is increasing research into therapies that target tau proteins and inflammation in the brain. In fact, inflammation and immune-targeting drugs now make up a growing portion of the Alzheimer's treatment pipeline. Researchers are also studying whether treating Alzheimer's before noticeable symptoms develop could delay or prevent cognitive decline, as well as whether combining treatments that target different parts of the disease could be more effective. Overall, the goal of these next-generation thearapies is to move beyond simply managing symptoms and find treatment that can slower or prevent the underlying changes happening in the brain.
